```python
# Lọc hóa đơn nước Đức
basket_germany = (
    df_retail[df_retail['Country'] == 'Germany']
    .groupby(['InvoiceNo', 'Description'])['Quantity']
    .sum().unstack().reset_index().fillna(0)
    .set_index('InvoiceNo')
    .applymap(encode_units)
)

if 'POSTAGE' in basket_germany.columns:
    basket_germany.drop('POSTAGE', inplace=True, axis=1)

# Khai thác tập phổ biến với min_sup = 5%
itemsets_germany = apriori(basket_germany, min_support=0.05, use_colnames=True)

# Sinh luật kết hợp với min_conf = 50%
rules_germany = association_rules(itemsets_germany, metric="confidence", min_threshold=0.50)

print(f"=== Kết quả cho nước Đức (Germany) - Số lượng luật: {len(rules_germany)} ===")
for idx, row in rules_germany.iterrows():
    print(f"{list(row['antecedents'])} ==> {list(row['consequents'])} | Sup: {row['support']:.4f}, Conf: {row['confidence']:.4f}")
```
