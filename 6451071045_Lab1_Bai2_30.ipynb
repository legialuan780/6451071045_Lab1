```python
import pandas as pd
import matplotlib.pyplot as plt
# Khai báo thông tin sinh viên
MSSV = "6451071045"
HO_TEN = "Lê Gia Luân"

print(f"Họ và tên: {HO_TEN}")
print(f"MSSV: {MSSV}")
# a) Đọc file và đặt lại tiêu đề các cột
df_tb = pd.read_excel('tb_tracking.xlsx', header=None)
df_tb.columns = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']

# Chuyển cột time về định dạng datetime
df_tb['time'] = pd.to_datetime(df_tb['time'])

# b) Lọc bỏ dữ liệu trước ngày 03/01/2018
df_filtered = df_tb[df_tb['time'] >= '2018-01-03'].copy()

# c) Bóc tách thời điểm đo đạc trong ngày tính theo phút (minutes)
df_filtered['minutes'] = (
    df_filtered['time'].dt.hour * 60 + 
    df_filtered['time'].dt.minute + 
    df_filtered['time'].dt.second / 60.0
)

# d) Vẽ đồ thị phân tán (scatter plots)
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

# Đồ thị Nồng độ CO
axes[0].scatter(df_filtered['minutes'], df_filtered['co_level'], alpha=0.6, color='red', s=12)
axes[0].set_title('Nồng độ khí CO theo phút trong ngày')
axes[0].set_xlabel('Minutes (Phút)')
axes[0].set_ylabel('CO Level')

# Đồ thị Độ ẩm
axes[1].scatter(df_filtered['minutes'], df_filtered['humidity'], alpha=0.6, color='blue', s=12)
axes[1].set_title('Độ ẩm theo phút trong ngày')
axes[1].set_xlabel('Minutes (Phút)')
axes[1].set_ylabel('Humidity (%)')

# Đồ thị Nhiệt độ
axes[2].scatter(df_filtered['minutes'], df_filtered['temperature'], alpha=0.6, color='green', s=12)
axes[2].set_title('Nhiệt độ theo phút trong ngày')
axes[2].set_xlabel('Minutes (Phút)')
axes[2].set_ylabel('Temperature (°C)')

plt.tight_layout()
plt.show()
```
