```python
import pandas as pd
import numpy as np

# Đọc dữ liệu Bank Marketing
df_bank = pd.read_csv('bank.csv', sep=';')

# 1. Xem thông tin cơ bản
print("=== Thông tin bộ dữ liệu Bank Marketing ===")
df_bank.info()

# 2. Kiểm tra các giá trị 'unknown' (thiếu thông tin)
unknown_counts = (df_bank == 'unknown').sum()
print("\n=== Số lượng giá trị 'unknown' theo từng thuộc tính ===")
print(unknown_counts[unknown_counts > 0])

# 3. Mã hóa thuộc tính nhị phân (Binary Encoding)
binary_cols = ['default', 'housing', 'loan', 'y']
for col in binary_cols:
    df_bank[col] = df_bank[col].map({'yes': 1, 'no': 0})

# 4. Tiền xử lý dữ liệu 'unknown'
# Điền Mốt cho 'job' và 'education'
df_bank['job'] = df_bank['job'].replace('unknown', df_bank['job'].mode()[0])
df_bank['education'] = df_bank['education'].replace('unknown', df_bank['education'].mode()[0])

# 5. Chia nhóm các thuộc tính định lượng liên tục (Binning)
df_bank['age_group'] = pd.qcut(df_bank['age'], q=5, labels=[0, 1, 2, 3, 4])
df_bank['duration_group'] = pd.qcut(df_bank['duration'], q=5, labels=[0, 1, 2, 3, 4])

# 6. Biến đổi thuộc tính dạng Categorical bằng One-Hot Encoding
df_bank_encoded = pd.get_dummies(
    df_bank, 
    columns=['job', 'marital', 'education', 'contact', 'poutcome', 'month'], 
    drop_first=True
)

print("\n=== Kích thước dữ liệu sau khi tiền xử lý ===", df_bank_encoded.shape)
print(df_bank_encoded.head())
```
